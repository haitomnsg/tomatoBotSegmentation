```python
# Import YOLO from ultralytics
from ultralytics import YOLO
```

```python
# Load a COCO-pretrained YOLOv8n segmentation model
model = YOLO("yolov8n-seg.pt")
```

```python
# Train the segmentation model
results = model.train(
    data="datasetFinal/data.yaml",  # Path to dataset configuration file
    epochs=10,  # Number of training epochs
    imgsz=640,  # Input image size
    batch=8,  # Batch size (adjust based on GPU memory)
    workers=2,  # Number of data loading workers
    device=0,  # Use GPU (NVIDIA)

    # Learning Rate & Optimization
    lr0=0.001,  # Initial learning rate
    lrf=0.1,  # Final learning rate factor
    momentum=0.937,  # Momentum for SGD
    weight_decay=0.0005,  # Weight decay for regularization

    # Data Augmentations
    hsv_h=0.015,  # Hue augmentation
    hsv_s=0.7,  # Saturation augmentation
    hsv_v=0.4,  # Value augmentation
    scale=0.5,  # Scaling augmentation
    translate=0.2,  # Translation augmentation
    shear=0.2,  # Shearing augmentation
    flipud=0.1,  # Vertical flip probability
    fliplr=0.5,  # Horizontal flip probability

    # Segmentation-specific parameters
    task='segment',  # Specify segmentation task
    mask_ratio=4,  # Mask downsampling ratio
    overlap_mask=True,  # Allow overlapping masks
    retina_masks=True  # Use high-resolution masks
)
```

```python
# Test the segmentation model on an image
model = YOLO("runs/segment/train/weights/best.pt")  # Load the best trained model
results = model("../test/Tomato_Image_1.jpg")  # Test on a sample image
results[0].show()  # Display the segmentation results
```
